```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/tweets_features_with_topics.csv")
```

```python
# Distribución de temas global
plt.figure(figsize=(10,5))
df["topic_id"].value_counts().sort_index().plot(kind="bar")
plt.title("Distribución de temas (LDA)")
plt.xlabel("Topic ID")
plt.ylabel("Número de tuits")
plt.show()
```

```python
# Distribución de temas por candidato
plt.figure(figsize=(14,7))
sns.countplot(data=df, x="topic_id", hue="candidate", palette="viridis")
plt.title("Temas por candidato")
plt.show()
```
